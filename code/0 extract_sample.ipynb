```python
import duckdb
import pandas as pd
import json
from pathlib import Path

print("Initializing DuckDB and generating complete TalkTalk sample extracts...")

# Connect to DuckDB
con = duckdb.connect(database=':memory:')
DATA = Path('../data')   # all input files are next to this notebook
OUT  = Path('../out')   # outputs land here too

# ==========================================
# 1. ESTABLISH THE ANCHOR SAMPLE
# ==========================================
# Extract 50 random customer IDs to act as our universal filter
con.execute(f"""
    CREATE TEMP TABLE sample_ids AS
    SELECT unique_customer_identifier 
    FROM read_parquet('{DATA / 'customer_info.parquet'}') 
    USING SAMPLE 50 ROWS;
""")

# ==========================================
# 2. EXTRACT DATA MAINTAINING INTEGRITY
# ==========================================
print("Extracting customer_info sample...")
customer_sample_df = con.execute(f"""
    SELECT * FROM read_parquet('{DATA / 'customer_info.parquet'}')
    WHERE unique_customer_identifier IN (SELECT unique_customer_identifier FROM sample_ids)
""").df()
customer_sample_df.to_csv(OUT/'lovable_sample_customer_info.csv', index=False)

print("Extracting usage sample...")
usage_sample_df = con.execute(f"""
    SELECT * FROM read_parquet('{DATA / 'usage.parquet'}')
    WHERE unique_customer_identifier IN (SELECT unique_customer_identifier FROM sample_ids)
""").df()
usage_sample_df.to_csv(OUT / 'lovable_sample_usage.csv', index=False)

print("Extracting calls sample...")
# Note: Using sample_size=-1 to bypass CSV type-guessing errors
calls_sample_df = con.execute(f"""
    SELECT * FROM read_csv_auto('{DATA / 'calls.csv'}' , sample_size=-1, ignore_errors=true)
    WHERE unique_customer_identifier IN (SELECT unique_customer_identifier FROM sample_ids)
""").df()
calls_sample_df.to_csv(OUT / 'lovable_sample_calls.csv', index=False)

print("Extracting cease sample...")
cease_sample_df = con.execute(f"""
    SELECT * FROM read_csv_auto('{DATA / 'cease.csv'}', sample_size=-1, ignore_errors=true)
    WHERE unique_customer_identifier IN (SELECT unique_customer_identifier FROM sample_ids)
""").df()
cease_sample_df.to_csv(OUT / 'lovable_sample_cease.csv', index=False)

# ==========================================
# 3. EXTRACT FULL SCHEMAS FOR LOVABLE
# ==========================================
print("Extracting schema definitions for all tables...")
schemas = {
    "customer_info": "DESCRIBE SELECT * FROM read_parquet(DATA / 'customer_info.parquet')",
    "usage": "DESCRIBE SELECT * FROM read_parquet(DATA / 'usage.parquet')",
    "calls": "DESCRIBE SELECT * FROM read_csv_auto(DATA / 'calls.csv', sample_size=-1, ignore_errors=true)",
    "cease": "DESCRIBE SELECT * FROM read_csv_auto(DATA / 'cease.csv', sample_size=-1, ignore_errors=true)"
}

schema_definition = {}
for table_name, query in schemas.items():
    schemas = {
        "customer_info": f"DESCRIBE SELECT * FROM read_parquet('{DATA / 'customer_info.parquet'}')",
        "usage": f"DESCRIBE SELECT * FROM read_parquet('{DATA / 'usage.parquet'}')",
        "calls": f"DESCRIBE SELECT * FROM read_csv_auto('{DATA / 'calls.csv'}', sample_size=-1, ignore_errors=true)",
        "cease": f"DESCRIBE SELECT * FROM read_csv_auto('{DATA / 'cease.csv'}', sample_size=-1, ignore_errors=true)"
    }
    for table_name, query in schemas.items():
        schema_df = con.execute(query).df()
        schema_definition[f"Table: {table_name}"] = {
            row['column_name']: row['column_type'] 
            for _, row in schema_df.iterrows()
        }

# Save the schema to a text file for your master prompt
with open(OUT / 'lovable_full_schema_types.txt', 'w') as f:
    f.write("TALKTALK COMPLETE DATA MODEL SCHEMA:\n\n")
    f.write(json.dumps(schema_definition, indent=4))

print("✅ Complete! Check your out folder for the four CSVs and the Full Schema text file.")
```
